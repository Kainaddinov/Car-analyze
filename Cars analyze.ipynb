# **Шаг 3: Clean Data**

## 1. Меняю названия колонок

```python
df.columns
```

```python
df = df.rename(columns={
    'Make': 'make',
    'Model': 'model',
    'Type': 'type',
    'Origin': 'origin',
    'DriveTrain': 'drive_train',
    'MSRP': 'msrp',
    'Invoice': 'invoice',
    'EngineSize': 'engine_size',
    'Cylinders': 'cylinders',
    'Horsepower': 'horse_power',
    'MPG_City': 'mpg_city',
    'MPG_Highway': 'mpg_highway',
    'Weight': 'weight',
    'Wheelbase': 'wheel_base',
    'Length': 'length'
})

# мне удобнее работать с маленькими буквами
```

## Поменял название колонок

```python
df.columns
```

## 2. Меняею типы данных

```python
df.head(3)
```

```python
# тут уже видно что msrp, invoice поменять тип данных

df.dtypes
```

## Убираю для начало $ и ,

```python
df['msrp'] = df['msrp'].str.replace('$', '')
df['msrp'] = df['msrp'].str.replace(',', '')
```

```python
df['invoice'] = df['invoice'].str.replace('$', '')
df['invoice'] = df['invoice'].str.replace(',', '')
```

```python
df.head(2)
```

## Теперь уже меняем object в двух столбцах на float

```python
df.dtypes
```

```python
df['msrp'] = df['msrp'].astype(float)
```

```python
df['invoice'] = df['invoice'].astype(float)
```

## Успешно поменял типы данных в колонках 'msrp', 'invoice'

```python
df.dtypes
```

# 2. 🏎️ Двигатель и мощность

## 2.1. Связана ли мощность с размером двигателя?

```python
plt.figure(figsize=(8, 5))


sns.scatterplot(df, x='engine_size', y='horse_power', s=100)

plt.title('Взаимосвязь между мощностью и размером двигателя')
plt.grid(True, linestyle='--', alpha=0.6)

plt.show()
```

### Ответ да связано. На диаграмме четко видно, чем больше размер двигателя, тем больше увеличивается мощность.

# 2.2. Увеличивается ли цена вместе с мощностью?

```python
plt.figure(figsize=(8, 5.5))

sns.regplot(df, x='horse_power', y='msrp', scatter_kws={'s': 60, 'alpha': 0.6}, line_kws={'color': 'red'})
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

### Исходя из диаграммы, можно сделать вывод что да, чем больше лошадиных сил у машины, тем больше его и стоимость.
